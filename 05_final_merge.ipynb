### Final Dataset Construction (Tourism + Weather)

This notebook merges tourism demand data with climate features to create the final modeling dataset.

### Objective:
- Align tourism demand with corresponding weather conditions
- Construct a unified dataset for machine learning models

### Join Strategy:
- Left join on (`prefecture`, `year`, `month`)
- Tourism data is treated as the target dataset

### Output:
data/final_tourism_weather_2016_2024.csv

```python
import pandas as pd
import os
```

```python
weather_path = "data/weather_monthly_2007_2025_new.csv"
tourism_path = "data/tourism_monthly/tourism_monthly_2007_2025.csv"

weather = pd.read_csv(weather_path)
tourism = pd.read_csv(tourism_path)

weather.shape, tourism.shape
weather.info(), tourism.info()
```

```python
tourism.columns = tourism.columns.str.strip().str.lower()
weather.columns = weather.columns.str.strip().str.lower()
```

```python
# Left join to preserve all tourism observations (target variable)
final_dataset = tourism.merge(
    weather , 
    on = ["prefecture","year","month"],
    how="left"
)
final_dataset.head(), final_dataset.shape
    
```

```python
final_dataset.isna().sum()
```

```python
tourism["year"].unique(), weather["year"].unique()
```

```python
final_dataset.duplicated(
    subset=["prefecture", "year", "month"]
).sum()
```

```python
final_path = "data/final_tourism_weather_2007_2025.csv"
final_dataset.to_csv(final_path, index = False)
final_path
```
