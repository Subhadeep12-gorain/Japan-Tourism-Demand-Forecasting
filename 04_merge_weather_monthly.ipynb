###  Weather Data Consolidation (Monthly Features)

This notebook combines all prefecture-level monthly weather datasets into a single unified dataset.

### Purpose:
- Enable global modeling across all prefectures
- Ensure consistent feature schema
- Prepare climate features for integration with tourism demand

### Output:
data/weather_monthly_2016_2024.csv

```python
import pandas as pd
import os
```

```python
WEATHER_DIR = "data/weather_monthly_new_1"
OUTPUT_DIR = "data"

os.makedirs(OUTPUT_DIR,exist_ok=True)
```

```python
final = []

EXPECTED_COLS = {
    "prefecture",
    "year",
    "month",
    "temp_avg",
    "temp_max",
    "temp_min",
    "precip_sum",
}

for fname in sorted(os.listdir(WEATHER_DIR)):

    if not fname.endswith(".csv"):
        continue

    path = os.path.join(WEATHER_DIR, fname)
    df = pd.read_csv(path)

    # Normalize column names
    df.columns = df.columns.str.strip().str.lower()

    # Schema validation
    missing = EXPECTED_COLS - set(df.columns)
    if missing:
        raise ValueError(f"{fname} missing columns: {missing}")

    final.append(df)
```

```python
weather_monthly_all = pd.concat(final , ignore_index= True)
```

```python
weather_monthly_all.duplicated(
    subset=["prefecture", "year", "month"]
).sum()
```

```python
weather_path = os.path.join(OUTPUT_DIR,"weather_monthly_2007_2025_new.csv")
weather_monthly_all.to_csv(weather_path, index=False)
weather_monthly_all
```
